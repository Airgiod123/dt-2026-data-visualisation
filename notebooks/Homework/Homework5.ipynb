Dataset on English kings from 1066 to 1485, since that marks the end of the Plantagenet dynasty on the English throne.
And the question for this dataset is: How did the length of English kings' reigns vary between the Norman Conquest all the way until the end of Richard III’s reign?

```python
import pandas as pd
df = pd.read_csv("kings.csv")
df.head()
df.describe()
# This is just to give a definition to the dataset and also a small statistics table
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="King/reign period", y="Reign length", legend=False)
plt.title("Length of English Reign Periods")
plt.ylabel("Reign length")
plt.xlabel("King/reign period")
plt.xticks(rotation=90)
plt.show()
```

The bar chart, gives an overview on how long each king ruled, and can mark out kings with short reigns so later research can be made to understand why theirs were shorter, for example Edward V the chart marks at 0 years because he didn't reign for an entire year but months so this is a limitation of the dataset since it only gives years and not the exact days and months. This chart can help see that long lived kings were not a minority but that a good amount of kings had reigns longer than 20 years.

```python
plt.scatter(df["Start year"], df["Reign length"])
plt.title("Reign Length by Start Year")
plt.ylabel("Reign length")
plt.xlabel("Start year")
plt.show()
```

The scatter chart doesn't show any signs of major instability for centuries, its a normal chart reigns but at the 15 century, it becomes more compact and by that one the realization that something must have occurred then and we know it was the War of The Roses. The chart also helps see the kings who had very long reigns like Henry VI during the century that would mark an end to the Plantagenet dynasty.

```python
df[["Reign length"]].boxplot(figsize=(8, 6))
plt.title("Distribution of Reign Lengths")
plt.ylabel("Reign length")
plt.show()
```

The median reign length is 18 years, meaning that half of the recorded periods fall below this value and half above it. The lengths range from zero to 56 years. This shows that some reigns were much longer than others. The zero means that a reign started and ended in the same year, not that it did not happen. The reigns of kings like Edward III and Henry III help extend the upper part of the chart.
