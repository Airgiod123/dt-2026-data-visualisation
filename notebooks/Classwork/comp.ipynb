Chart of who puts in the most effort

```python
import pandas as pd 
df = pd.read_csv("competencies.csv")
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
avg_practice= df.groupby("Student")["Hours_Studied"].mean()
avg_practice.plot(kind="bar")
plt.ylabel("Hours Studied")
plt.title ("Average hours studied per student")
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Hours Studied by their Grades")
plt.ylabel("Grades")
plt.xlabel("Hours Studied")
```

```python
plt.Figure(figsize=(8,6))
df[["Grade", "Hours_Studied"]].boxplot()
plt.title("Boxplot of grade and study hour consistency")
plt.ylabel("Grades")
plt.show
```
