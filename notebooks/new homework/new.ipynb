```python
import urllib.request
url = "https://www.gutenberg.org/cache/epub/174/pg174.txt"
text = urllib.request.urlopen(url).read().decode('utf-8')
words = text.split()
print(f"The text has {len(words)} words.")
print(f"The first 10 words are: {words[:10]}")
print("The first letter of each of the first 10 words is:")
for word in words[:10]:
    print(word[0])
```

I used urlib.request so the code could read the txt file I gave, then that it could decode and actively use it. Then I use split to make the text into words and then the first print to say how many words are in the text. After that words[:10] is used to get the first 10 words and then for the last print a for loop so it could print the first letter of the first 10 words. 

```python
import urllib.request
from wordcloud import WordCloud
url = "https://www.gutenberg.org/cache/epub/1184/pg1184.txt"
text = urllib.request.urlopen(url).read().decode('utf-8')
stopwords_list = urllib.request.urlopen("https://raw.githubusercontent.com/stopwords-iso/stopwords-en/master/stopwords-en.txt").read().decode("utf-8").split("\n")
def color_function(word, font_size, position, orientation, random_state=None, **additional_args):
    if len(word) < 5:
        return "red"
    else:
        return "yellow"
cloud = WordCloud(stopwords=stopwords_list, background_color="black", width=800, height=400)
cloud.generate_from_text(text).recolor(color_func=color_function)
cloud.to_image()
```

I used The Count of Monte Cristo as the text from which to form the world cloud. I used the stopword list to filter out common words, so names and titles would be better to see. Count and Monte Cristo are very noticable since it's how Edmond is refered to in the book commonly, there are also many names like Albert, Morrel and Valentine since they are characters it's normal they are much bigger than many other words. Within the code I made it so words that are 4 or less letters red and everything longer is yellow and made the background black so reading would be easier on the eyes, for that I used an if and else statement, the rest of it is the same as just reading the text execpt the wordcloud for that its given a definition of what it does with the text and then that definition is used to make the image.
